### Financial Trend Analysis: Microsoft, Tesla, and Apple (FY 2023-2025)
## 1. Introduction & Methodology
This report provides a comprehensive financial performance analysis of three major technology leaders: Microsoft (MSFT), Tesla (TSLA), and Apple (AAPL). The data was manually extracted from the official SEC 10-K filings for the fiscal years 2023 through 2025.

The primary objective is to evaluate year-over-year (YoY) growth trends and operational efficiency to provide a foundational dataset for the GFC AI-powered financial chatbot. The analysis was conducted using Python's Pandas library, focusing on key metrics such as revenue growth, net income margins, and asset turnover.

## 2. Key Findings & Strategic Insights
**Microsoft Corporation:** Sustained Scaling
Microsoft demonstrates a remarkably robust and consistent growth trajectory.

Revenue Performance: The company achieved a significant jump in total revenue, scaling from **$211.9B** in 2023 to **$281.7B** by 2025.

Profitability: Unlike many of its peers, Microsoft managed to grow its net income proportionally, indicating strong operational leverage and successful integration of AI-driven cloud services.

**Tesla, Inc.:** Margin Compression & Market Headwinds
Tesla’s financial data reveals a more volatile story, highlighting the challenges within the EV sector.

Profitability Crunch: While revenue remained relatively stable, Net Income saw a sharp decline from **$14.9B** in 2023 to roughly **$3.8B** in 2025. This suggests significant margin compression, likely due to aggressive price cuts, increased R&D spending, or rising production costs.

Strategic Note: For the AI chatbot, this data point is crucial for advising users on "risk-adjusted" growth expectations versus pure revenue volume.

**Apple Inc.:** The Efficiency Benchmark
Apple continues to operate with unparalleled scale, though its growth has shifted toward a "stability and efficiency" phase.

Top-Line Volume: With a total revenue exceeding **$416B** in 2025, Apple remains the volume leader.

Asset Optimization: Despite a massive asset base, Apple maintains a high Asset Turnover ratio, showcasing its ability to generate significant sales from its resources.

## 3. Conclusion for AI Chatbot Integration
The calculated metrics (Growth %, Asset Ratios) are now structured into a "machine-readable" format. This allows the GFC chatbot to:

Compare relative performance between these giants accurately.

Identify red flags, such as Tesla's declining net income despite stable sales.

Project future trends based on the historical volatility observed in this three-year window.

```python
def start_gfc_advanced_bot():
    print("--- GFC Financial Chatbot Prototype ---")
    print("Ask me about revenue, net income trends, or the leader of 2025.")
    print("(Type 'exit' to end the session)\n")

    while True:
        # Get input and clean it up
        query = input("Enter your query: ").lower().strip()
        if query == 'exit':
            print("Analysis complete. METU IE excellence delivered! See you soon.")
            break

        # --- 1. ENTITY EXTRACTION ---
        target_company = None
        for c in ['microsoft', 'tesla', 'apple']:
            if c in query: target_company = c.capitalize()

        target_year = None
        for y in ['2023', '2024', '2025']:
            if y in query: target_year = int(y)

        # Metric mapping (mapping common words to your DataFrame columns)
        metric = 'Total Revenue' if any(x in query for x in ['revenue', 'sales', 'gelir']) else 'Net Income'
        metric_label = "Total Revenue" if metric == 'Total Revenue' else "Net Income"

        # --- 2. LOGIC FLOW (IF-ELSE STRUCTURE) ---

        # A. Extremes (Highest / Lowest)
        if any(x in query for x in ['max', 'highest', 'most', 'best']):
            sub_df = df[df['Year'] == target_year] if target_year else df
            res = sub_df.loc[sub_df[metric].idxmax()]
            print(f"🤖 [ANALYSIS]: The highest {metric_label} in {target_year if target_year else 'the recorded period'} belongs to {res['Company']} (${res[metric]:,.0f})\n")

        elif any(x in query for x in ['min', 'lowest', 'least', 'worst']):
            sub_df = df[df['Year'] == target_year] if target_year else df
            res = sub_df.loc[sub_df[metric].idxmin()]
            print(f"🤖 [ANALYSIS]: The lowest {metric_label} in {target_year if target_year else 'the recorded period'} belongs to {res['Company']} (${res[metric]:,.0f})\n")

        # B. Trend & Change Analysis (2025 Growth Check)
        elif any(x in query for x in ['change', 'diff', 'trend', 'growth']):
            if target_company:
                growth_col = 'Revenue Growth (%)' if metric == 'Total Revenue' else 'Net Income Growth (%)'
                change = df[(df['Company'] == target_company) & (df['Year'] == 2025)][growth_col].values[0]
                status = "increased 📈" if change > 0 else "decreased 📉"
                print(f"🤖 [TREND]: {target_company}'s {metric_label} has {status} by {abs(change):.2f}% in 2025 compared to the previous year.\n")
            else:
                print("🤖 Please specify which company you'd like to analyze for trends.\n")

        # C. Specific Data Point (Company + Year)
        elif target_company and target_year:
            val = df[(df['Company'] == target_company) & (df['Year'] == target_year)][metric].values[0]
            print(f"🤖 [DATA]: {target_company} | {target_year} | {metric_label}: ${val:,.0f}\n")

        # D. General Company Overview
        elif target_company:
            print(f"🤖 [SUMMARY]: Financial history for {target_company} ({metric_label}):")
            summary = df[df['Company'] == target_company][['Year', metric]]
            print(summary.to_string(index=False), "\n")

        # E. Error Handling
        else:
            print("🤖 Please try asking about a company (Apple, Tesla, Microsoft), a year (2023-2025), or trends like 'highest revenue'.\n")

# Run the engine
start_gfc_advanced_bot()
```
